###AUTOLOADER###

```python
from pyspark.sql.functions import *
```

**Streaming** **Dataframe**

```python
df = spark.readStream.format("cloudFiles") \
        .option("cloudFiles.format", 'parquet') \
        .option("cloudFiles.schemaLocation", 'abfss://mycontainer@storagemoderndbfff.dfs.core.windows.net/autosink/check') \
        .load("abfss://mycontainer@storagemoderndbfff.dfs.core.windows.net/autosource")
```

```python
df.writeStream.format('parquet')\
        .option("checkpointLocation", 'abfss://mycontainer@storagemoderndbfff.dfs.core.windows.net/autosink/check')\
        .trigger(processingTime='10 seconds')\
        .start('abfss://mycontainer@storagemoderndbfff.dfs.core.windows.net/autosink/data')
```
