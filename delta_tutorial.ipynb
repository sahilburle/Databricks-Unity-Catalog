**Delta Table**

```python
%sql
CREATE TABLE man_cata.man_schema.deltatbl
(
    id INT,
    name STRING,
    city STRING
)
USING DELTA
LOCATION 'abfss://mycontainer@storagemoderndbfff.dfs.core.windows.net/deltalake/deltatbl'
```

```python
%sql
ALTER TABLE man_cata.man_schema.deltatbl SET TBLPROPERTIES ('delta.enableDeletionVectors' = 'false');
```

```python
%sql
INSERT INTO man_cata.man_schema.deltatbl
VALUES
(1, 'aa', 'delhi'),
(2, 'bb', 'london'),
(3, 'cc', 'sydney')
```

```python
%sql
DESCRIBE EXTENDED man_cata.man_schema.deltatbl
```

```python
%sql
SELECT * FROM man_cata.man_schema.deltatbl
```

**Updates in Delta Tables**

```python
%sql
UPDATE man_cata.man_schema.deltatbl
SET city = 'toronto'
WHERE id = 1
```

```python
%sql
DESCRIBE HISTORY man_cata.man_schema.deltatbl
```

**Time Travel**

```python
%sql
RESTORE man_cata.man_schema.deltatbl TO VERSION AS OF 2
```

**Deletion Vector**

```python
%sql
CREATE TABLE man_cata.man_schema.deltatable2
(
    id INT,
    name STRING,
    city STRING
)
USING DELTA
LOCATION 'abfss://mycontainer@storagemoderndbfff.dfs.core.windows.net/deltalake/deltatable2'
```

```python
%sql
ALTER TABLE man_cata.man_schema.deltatbl SET TBLPROPERTIES ('delta.enableDeletionVectors' = 'true');
```

```python
%sql
INSERT INTO man_cata.man_schema.deltatable2
VALUES
(1, 'aa', 'delhi'),
(2, 'bb', 'london'),
(3, 'cc', 'sydney')
```

**Updates in Deletion Vector Table**

```python
%sql
UPDATE man_cata.man_schema.deltatable2
SET city = 'umred'
WHERE id = 1
```

```python
%sql
DESCRIBE HISTORY man_cata.man_schema.deltatable2
```

### **Optimization in Delta Tables** ###

### Deep Clone vs Shallow Clone ###

**Deep Clone**

```python
%sql
CREATE TABLE man_cata.man_schema.deepclonetbl
DEEP CLONE man_cata.man_schema.deltatbl
```

```python
%sql
SELECT * FROM  man_cata.man_schema.deepclonetbl
```

```python
%sql
SELECT * FROM  man_cata.man_schema.deltatbl
```

```python
%sql
DESCRIBE HISTORY man_cata.man_schema.deepclonetbl
```

**Shallow Clone**

```python
%sql
CREATE TABLE man_cata.man_schema.shallowtbl
SHALLOW CLONE man_cata.man_schema.man_table
```
